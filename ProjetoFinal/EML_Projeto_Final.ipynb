```python
# 1. Importação das bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

# 2. Carregamento dos dados locais do Colab (California Housing)
df = pd.read_csv("/content/sample_data/california_housing_train.csv")

print("Bibliotecas e dados de casas da Califórnia carregados com sucesso localmente.")
```

```python
# 3. Inspeção inicial
print("Dimensões:", df.shape)
display(df.head())

print("\nTipos de dados:")
display(df.dtypes)

print("\nValores ausentes:")
display(df.isnull().sum().sort_values(ascending=False).head(10))

print("\nEstatísticas descritivas:")
display(df.describe())
```

```python
# Seleção de atributos ajustados para o dataset da Califórnia
features = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income"
]

X = df[features]
y = df["median_house_value"]

# Separação dos dados
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Modelos
modelos = {
    "Regressão Linear": LinearRegression(),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

resultados = []

for nome, modelo in modelos.items():
    pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", modelo)
    ])

    pipeline.fit(X_train, y_train)
    previsoes = pipeline.predict(X_test)

    resultados.append({
        "Modelo": nome,
        "R²": r2_score(y_test, previsoes),
        "MAE": mean_absolute_error(y_test, previsoes),
        "RMSE": np.sqrt(mean_squared_error(y_test, previsoes))
    })

comparacao = pd.DataFrame(resultados)
display(comparacao.round(3))
```
